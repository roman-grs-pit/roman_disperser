# Example of using Synphot to get fluxes

This is based on the Roman Documentation [here](https://roman-docs.stsci.edu/simulation-tools/additional-simulation-tools/synphot-for-roman).

Uses the `roman_disperser.refdata` module for the F158 bandpass and stellar templates
(no stsynphot or PYSYN_CDBS required).

```python
from astropy import units as u
from astropy.time import Time
from matplotlib import pyplot as plt
import numpy as np

import synphot as syn
from roman_disperser import refdata
```

```python
wfi_f158 = refdata.get_f158_band()
print('WFI F158:')
print(f'\tBandwidth: {wfi_f158.photbw():.5f}')
print(f'\tPivot wavelength: {wfi_f158.pivot():.5f}')
print(f'\tFWHM: {wfi_f158.fwhm():.5f}') 
print(f'\tThroughput at 1.58 um: {wfi_f158(1.58 * u.micron):.4f}')
```

```python
# Retrieve the spectrum of the star
# This is a G0V star from the Bruzual Atlas
star_spec = refdata.get_template("g0v")
 
# Renormalize the spectrum to have a F158 magnitude of 20 ABmag
norm_spec = star_spec.normalize(20 * u.ABmag, band=wfi_f158)
 
# Compute the flux density arrays
wavelengths = np.arange(0.5, 2.5, 0.001) * u.micron
norm_fluxes = norm_spec(wavelengths, flux_unit=syn.units.FLAM)
 
# Plot the spectra
fig, ax = plt.subplots()
ax.plot(wavelengths, norm_fluxes,
        color='blue',
        lw=2,
        label='Normalized Spectrum, F158 = 20 ABmag')

ax.legend()

ax.set_ylabel('Flux Density (erg / s / cm^2 / Angstrom)')
ax.set_xlabel(r'Wavelength ($\mu$m)')

plt.show()
```
